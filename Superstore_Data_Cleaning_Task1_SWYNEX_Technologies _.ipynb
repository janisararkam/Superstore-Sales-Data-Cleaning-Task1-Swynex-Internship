```python
df["Order Date"] = pd.to_datetime(df["Order Date"], format="%d/%m/%Y")
df["Ship Date"] = pd.to_datetime(df["Ship Date"], format="%d/%m/%Y")
df["Postal Code"] = df["Postal Code"].astype("Int64")
```

```python
df.loc[df["Postal Code"].isna(), "Postal Code"] = 5401
```

```python
df.isnull().sum()
```

```python
df.rename(columns={"Row ID": "Row_ID"}, inplace=True)
df.rename(columns={"Order ID": "Order_ID"}, inplace=True)
df.rename(columns={"Order Date": "Order_Date"}, inplace=True)
df.rename(columns={"Ship Date": "Ship_Date"}, inplace=True)
df.rename(columns={"Ship Mode": "Ship_Mode"}, inplace=True)
df.rename(columns={"Customer ID": "Customer_ID"}, inplace=True)
df.rename(columns={"Customer Name": "Customer_Name"}, inplace=True)
df.rename(columns={"Postal Code": "Postal_Code"}, inplace=True)
df.rename(columns={"Product ID": "Product_ID"}, inplace=True)
df.rename(columns={"Sub-Category": "Sub_Category"}, inplace=True)
df.rename(columns={"Product Name": "Product_Name"}, inplace=True)
```

```python
#Categorical columns — unexpected values
for col in ["Ship_Mode", "Segment", "Country", "Region", "Category", "Sub_Category"]:
    print(f"\n{col}:")
    print(df[col].unique())
```

```python
#. ID columns — invalid values
print("Row_ID <= 0:", (df["Row_ID"] <= 0).sum())
print("Missing Order_ID:", df["Order_ID"].isnull().sum())
print("Missing Customer_ID:", df["Customer_ID"].isnull().sum())
print("Missing Product_ID:", df["Product_ID"].isnull().sum())
```

```python
#3. Postal Code — invalid values
print("Postal_Code <= 0:", (df["Postal_Code"] <= 0).sum())
print("Postal_Code missing:", df["Postal_Code"].isnull().sum())
```

```python
#Sales — zero/negative values
print("Sales <= 0:", (df["Sales"] <= 0).sum())
```

```python
#Sales — statistical outliers (IQR)
Q1 = df["Sales"].quantile(0.25)
Q3 = df["Sales"].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print("Lower limit:", lower)
print("Upper limit:", upper)
print("Outliers:", ((df["Sales"] < lower) | (df["Sales"] > upper)).sum())
```

```python
df[df["Sales"] > upper]["Sales"].describe()
```

```python
df[df["Sales"] > upper][["Sales", "Category", "Sub_Category", "Product_Name"]].sort_values("Sales", ascending=False).head(20)
```
